```python
# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(temparatures_in_celsious):
    return (temparatures_in_celsious * 9/5) + 32
```

```python
# Specify the temperature threshold in Celsius
threshold = 30
```

```python
temparatures_in_celsious =[27,28,30,31,32]
print(temparatures_in_celsious)
```

```python
# Convert and display each temperature
for temp in temparatures_in_celsious:
    temp_f = celsius_to_fahrenheit(temp)
    print(f"{temp}°C = {temp_f:.2f}°F")
```

```python
# Compare each temperature with the threshold
for celsius in temparatures_in_celsious:
    print(f"Checking temperature {celsius}°C")
    if celsius > threshold:
        print("  The temperature is above the threshold.")
    elif celsius < threshold:
        print("  The temperature is below the threshold.")
    else:
        print("  The temperature is equal to the threshold.")
    print()
```
